```python
import matplotlib.pyplot as plt
from astropy.io import fits
import os
import numpy as np
```

```python
file_name_first = '../playing_with_scopesim/first.fits'
file_name_second = '../playing_with_scopesim/second.fits'
```

```python
# does file exist?
if not os.path.exists(file_name_first):
    raise FileNotFoundError(f"File {file_name_first} does not exist")
```

```python
with fits.open(file_name_first) as hdul:
    data_first = hdul[1].data

with fits.open(file_name_second) as hdul:
    data_second = hdul[1].data
```

```python
def info_array(array_pass):

    print("--------------------------------")
    print("Shape:", array_pass.shape)
    print("Data type:", array_pass.dtype)
    print("Min:", np.min(array_pass))
    print("Max:", np.max(array_pass))
    print("Mean:", np.mean(array_pass))
    print("Std dev:", np.std(array_pass))
```

```python
info_array(data_first)
info_array(data_second)
```

```python
plt.imshow(data_first/data_second, origin='lower')
plt.colorbar()
plt.show()
```
